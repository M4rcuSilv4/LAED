b) Analise o tempo de execução desse algoritmo no pior caso.


**RESPOSTAS:**

A) No mellhor caso, onde a lista está previamente ordenada e o pivô é o ultimo elemento da lista, o tempo de execução é O(n), já que a partição teria custo linear, e as ordenações de um lado seria O(n-1), lado com os elementos menores que o pivô previamente ordenados, e do outro O(1), lado com os elementos maiores que o pivô.

B) Já no pior caso, a partição seria desequilibrada e ambos lados estariam completamente desordenados, po essa lógica como a etapa de ordenação seria a mais custosa, no fim o tem po seria quadrático, ou seja, O(n^2).

A terceira passagem garante que os elementos que foram movidos para fora da primeira região na segunda passagem, agora podem ser realocados corretamente, consolidando a ordenação.

# 4. Verificar se existem dois elementos iguais na matriz *M*.

# 5. Verificar se a matriz *M* contém duas linhas exatamente iguais.

```python
def tem_linhas_repetidas(matriz):
    linhas_vistas = set()

    for linha in matriz:
        linha_imutavel = tuple(linha)

        if linha_imutavel in linhas_vistas:
            return True

        linhas_vistas.add(linha_imutavel)

    return False


AM = [
    [1, 2, 3],
    [4, 5, 6],
    [1, 2, 3]
]

PM = [
    [1, 2, 3],
    [3, 2, 1],
    [4, 5, 6]
]

print(f"AM tem linhas iguais? {tem_linhas_repetidas(AM)}")
print(f"PM tem linhas iguais? {tem_linhas_repetidas(PM)}")
```
