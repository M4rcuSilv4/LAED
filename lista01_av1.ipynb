# **Função para um criar vetor aleatório**

```python
import random
from random import randint

def creat_vetor(t, lim_i,lim_s):
  vetor = [0] * t

  for i in range(t):
    vetor[i] = random.randint(lim_i, lim_s)

  return vetor
```

# 1. O maior número ímpar

```python
b = creat_vetor(10,1,100)
print(b)

maior_imp = 0
x = 0

for x in range(len(b)):
  if b[x]%2!=0 and b[x]>maior_imp:
    maior_imp = b[x]
  else:
    x+=1

if maior_imp == 0:
  print("não possui ímpares.")
else:
  print(f"O maior ímpar é: {maior_imp}")
```

# 7. Contando inversões

```python
novo = creat_vetor(10,1,100)
print("Vetor inicial: \n", novo)

cont= 0
n = len(novo)
i = 0

while i < n - 1:
  j = 0
  while j < n - 1 - i:
    if novo[j] > novo[j+1]:
      aux = novo[j]
      novo[j] = novo[j+1]
      novo[j+1] = aux
      cont+=1
    j += 1
  i += 1

print("Vetor ordenado: \n",novo)
print(f"Total de {cont} inversões")
```
